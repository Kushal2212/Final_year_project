```python


import sys
import os


def train():
    from src.train_model import train as run_train
    run_train()


def predict(img_path):
    from src.predict import predict as run_predict
    result = run_predict(img_path)
    print(f"\nDisease     : {result['disease'].upper()}")
    print(f"Nepali Name : {result['nepali_name']}")
    print(f"Confidence  : {result['confidence']}%")
    print(f"Severity    : {result['severity']}")
    print(f"Description : {result['description']}")
    print(f"Recommendation:\n  {result['recommendation']}")


def evaluate():
    from src.utils import evaluate_model
    BASE_DIR   = os.path.dirname(os.path.abspath(__file__))
    model_path = os.path.join(BASE_DIR, "models", "cardamom_disease_model.h5")
    test_dir   = os.path.join(BASE_DIR, "dataset", "test")
    evaluate_model(model_path, test_dir)


def web():
    sys.path.insert(0, os.path.join(os.path.dirname(__file__), "app"))
    from app.app import app
    print("\n🌿 Starting web server at http://127.0.0.1:5000\n")
    app.run(debug=True, host="0.0.0.0", port=5000)


def preprocess():
    from src.data_preprocessing import split_and_copy
    BASE_DIR = os.path.dirname(os.path.abspath(__file__))
    split_and_copy(
        os.path.join(BASE_DIR, "raw_images"),
        os.path.join(BASE_DIR, "dataset"),
    )


COMMANDS = {
    "train":      lambda args: train(),
    "predict":    lambda args: predict(args[0]) if args else print("Provide image path"),
    "evaluate":   lambda args: evaluate(),
    "web":        lambda args: web(),
    "preprocess": lambda args: preprocess(),
}

if __name__ == "__main__":
    if len(sys.argv) < 2 or sys.argv[1] not in COMMANDS:
        print(__doc__)
        sys.exit(0)
    cmd  = sys.argv[1]
    args = sys.argv[2:]
    COMMANDS[cmd](args)
```
